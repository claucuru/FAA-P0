# Estrategia de Particionado

## Validación Simple

```python
datos = Datos("../datasets/heart-copy.csv")

estrategia=EstrategiaParticionado.ValidacionSimple(5, 0.8)
estrategia.creaParticiones(datos, 1)

for i, particion in enumerate(estrategia.particiones):
    print(f"Particion {i + 1}")
    print("  Train:", particion.indicesTrain)
    print("  Test:", particion.indicesTest)
```

## Validación Cruzada

Probamos con un dataset de 10 columnas y con K=4 particiones (distribución no uniforme):

```python
datos = Datos("../datasets/heart-copy.csv")

estrategia=EstrategiaParticionado.ValidacionCruzada(4)
estrategia.creaParticiones(datos, 1)

for i, particion in enumerate(estrategia.particiones):
    print(f"Particion {i + 1}")
    print("  Train:", particion.indicesTrain)
    print("  Test:", particion.indicesTest)
```
